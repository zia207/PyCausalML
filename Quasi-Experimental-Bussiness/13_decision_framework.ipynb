Start from the **assignment mechanism**, not from your favorite estimator:

```
How was the intervention assigned?
│
├─ By a threshold on a continuous score?  ───────────────► RDD (Section 10)
│
├─ To some units at known time(s), others never?
│    ├─ Many treated & control units, parallel pre-trends ► DiD / event-study (4–5)
│    ├─ Staggered adoption, dynamic effects ─────────────► Callaway–Sant'Anna (5.2)
│    ├─ One/few treated units, many good donors ─────────► Synthetic control (8)
│    └─ One treated series + predictive control series ──► CausalImpact (7)
│
├─ To everyone at once (no controls)? ───────────────────► ITS (6) — and lower
│                                                           your confidence accordingly
├─ Self-selected by units?
│    ├─ Selection plausibly on OBSERVABLES, rich X ──────► PS / doubly robust / DML (9, 12)
│    ├─ Selection on UNOBSERVABLES, instrument exists ───► IV / encouragement design (11)
│    └─ Neither ─────────────────────────────────────────► be honest: descriptive only,
│                                                           and design a better rollout next time
└─ Need "for whom," not just "on average"? ──────────────► CATE / causal forests (12)
```

Complementary heuristics:

- **Triangulate.** When two designs with *different* assumptions (e.g., DiD and SCM) agree, the conclusion is far stronger than either alone.
- **Design beats estimation.** The highest-leverage move is often upstream: persuade the business to stagger a rollout, hold out a few markets, or randomize the *encouragement* even when treatment can't be randomized. Quasi-experiments are easiest to analyze when they are quasi-*designed*.
- **Effect size vs. decision threshold.** Frame results against the decision ("the program pays for itself above +$5k/week; our 95% CI is [6.1, 9.8]") rather than against zero.
