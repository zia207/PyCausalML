# 7. Policy Learners for Causal Inference

Policy learning extends causal inference from estimation to decision-making. Instead of just asking "what is the treatment effect?", policy learning asks "who should be treated?" and provides actionable treatment assignment rules.

## Overview

### From Estimation to Decision-Making

Traditional causal inference focuses on estimating treatment effects (ATE, ATT, CATE). Policy learning goes further by translating these effect estimates into **actionable treatment policies**.

The key insight is that not everyone benefits from treatment. Policy learning identifies **heterogeneous treatment effects** and creates rules to assign treatment to those who benefit most.

### The Policy Learning Problem

Given:
- Covariates $X$ (features describing units)
- Potential outcomes $Y(0)$ and $Y(1)$
- Treatment $T \in \{0, 1\}$

We want to learn a **policy function** $\pi(X): \mathcal{X} \rightarrow \{0, 1\}$ that maximizes expected outcome:

$$
\max_{\pi} \mathbb{E}[Y(\pi(X))]
$$

This is equivalent to:

$$
\pi(x) = \begin{cases}
1 & \text{if } \tau(x) > 0 \\
0 & \text{if } \tau(x) \leq 0
\end{cases}
$$

where $\tau(x) = \mathbb{E}[Y(1) - Y(0) \mid X=x]$ is the conditional average treatment effect (CATE).

### Business Setting

Policy learning is ideal when:

- **Targeting is the goal**: You need to decide who gets treatment
- **Heterogeneous effects exist**: Treatment effects vary across the population
- **Resource constraints**: Treatment is expensive or limited
- **Personalization needed**: Different groups should receive different treatments

Common business applications:

- **Marketing**: Which customers should receive a promotion?
- **Pricing**: Dynamic pricing based on customer characteristics
- **Credit**: Which loan applicants should be approved?
- **Healthcare**: Which patients should receive a specific treatment?
- **Education**: Which students should receive tutoring?
- **Retention**: Which customers should get retention offers?

## Key Concepts

### Deterministic vs. Stochastic Policies

- **Deterministic policy**: Assigns treatment with certainty (0 or 1)
- **Stochastic policy**: Assigns treatment with probability (e.g., 70% chance of treatment)

Most policy learning methods focus on deterministic policies for simplicity and interpretability.

### Policy Value

The **policy value** is the expected outcome under the policy:

$$
V(\pi) = \mathbb{E}[Y(\pi(X))]
$$

Since we never observe counterfactuals, we estimate this using doubly robust estimators or other causal methods.

### Regret

**Regret** is the difference between the optimal policy value and the actual policy value:

$$
\text{Regret} = V(\pi^*) - V(\pi)
$$

where $\pi^*$ is the oracle optimal policy (knowing true CATEs). Minimizing regret is the goal of policy learning.

### Policy Learning vs. CATE Estimation

| Aspect | CATE Estimation | Policy Learning |
|---|---|---|
| **Goal** | Estimate $\tau(x)$ for all $x$ | Learn treatment assignment rule |
| **Output** | Function or value per unit | Binary treatment assignment |
| **Focus** | Understanding heterogeneity | Decision-making |
| **Evaluation** | RMSE, bias | Policy value, regret |

CATE estimation is a stepping stone to policy learning, but policy learning has its own methods and evaluation metrics.

## Advantages of Policy Learning

### 1. Actionable Output

Instead of just reporting "the effect varies," policy learning tells you exactly who to treat:

- CATE: "Customer A has effect +$10, Customer B has effect -$5"
- Policy: "Treat Customer A, don't treat Customer B"

### 2. Resource Optimization

When treatment is expensive or limited, policy learning maximizes impact per unit of resource:

- Limited budget for marketing → target high-response customers
- Limited supply of a product → allocate to high-value customers

### 3. Robustness to Estimation Error

Policy trees and forests can be more robust to estimation error than point estimates of CATE:

- Small errors in CATE may flip treatment decisions at the margin
- Policy methods use regularization to avoid overfitting

### 4. Interpretability

Policy trees provide clear, interpretable rules:

- "Treat if age > 30 AND income > $50k"
- "Don't treat if risk score < 0.3"

This is much easier to communicate to stakeholders than complex CATE functions.

## Challenges

### 1. Estimation Error

Policy learning relies on estimated CATEs, which have error. This error can propagate to the policy:

- Overfitting: Policy performs well on training data but poorly on new data
- Need for careful validation and regularization

### 2. Off-Policy Evaluation

Evaluating a policy requires estimating counterfactual outcomes, which is challenging:

- Need to estimate what would have happened under the alternative treatment
- Requires causal assumptions (unconfoundedness)

### 3. Sample Efficiency

Learning good policies requires sufficient data:

- Need enough treated and control units in each region of covariate space
- Sparse data can lead to poor policies

### 4. Fairness Constraints

Optimal policies may have undesirable distributional effects:

- Policy might systematically disadvantage certain groups
- May need to add fairness constraints to the optimization
