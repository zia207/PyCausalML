# 8. Quasi-Experimental Methods for Causal Inference


## Overview

Quasi-experiments are research designs that aim to estimate causal effects without full randomization — they occupy the space between a controlled experiment (where you randomly assign treatment) and pure observation (where you simply watch what happens). In business settings, true randomized controlled trials (RCTs) are often infeasible due to ethical, practical, or financial constraints. Quasi-experimental methods provide a rigorous alternative when you cannot randomize but still need credible causal estimates.

### When to Use Quasi-Experiments

Quasi-experimental methods are particularly valuable in business contexts such as:

- **Policy evaluations**: Assessing the impact of company-wide initiatives (e.g., pricing changes, new return policies)
- **Geographic rollouts**: When new features or programs launch in select markets before nationwide expansion
- **Natural experiments**: Leveraging external shocks or events that create as-if random variation
- **Historical analysis**: Retrospective evaluation of past interventions when RCTs were not conducted

### The Key Challenge: Selection Bias

The fundamental problem in observational data is **selection bias** — treated and control groups differ systematically in ways that affect outcomes. Naive comparisons between treated and untreated units often reflect pre-existing differences rather than treatment effects.

For example, if you compare revenue at stores that adopted a new program versus those that didn't, the adopters were likely already higher-performing stores. The observed difference mixes the true program effect with this pre-existing advantage.

Quasi-experimental methods solve this by finding **structured variation** in treatment assignment that allows credible counterfactual construction. Each method makes a different identifying assumption suited to a different assignment mechanism.

### Methods Covered in This Section

This section introduces five core quasi-experimental methods, each suited to different business scenarios:

| Method | Chapter | Business Setting | Key Assumption |
|---|---|---|---|
| **Difference-in-Differences (DiD)** | 6.1 | Treatment assigned to some groups at known time points | Parallel trends: treated and control groups would have evolved similarly absent treatment |
| **Synthetic Control** | 6.2 | One or few treated units with a pool of potential donors | A weighted combination of donors reproduces the treated unit's pre-intervention path |
| **Regression Discontinuity (RDD)** | 6.3 | Treatment assigned by a threshold on a continuous score | Units just above/below the cutoff are as good as randomized |
| **Instrumental Variables (IV)** | 6.4 | Treatment self-selected with confounding, but an instrument exists | Instrument affects outcome only through treatment (exclusion) |
| **Bayesian quasi-experiments** | 6.5 | Posterior intervals for DiD, synthetic control, RDD, ITS, and IV, plus a linear CATE | The design's identifying assumption. Uncertainty is a Bayesian bootstrap, or a normal posterior for the linear CATE |

### The Decision Framework

Choosing the right quasi-experimental method starts from the **assignment mechanism** — how treatment was actually assigned in your setting:

```
How was the intervention assigned?
│
├─ By a threshold on a continuous score?  ───────────────► RDD (Chapter 6.3)
│
├─ To some units at known time(s), others never?
│    ├─ Many treated & control units, parallel pre-trends ► DiD (Chapter 6.1)
│    ├─ One/few treated units, many good donors ─────────► Synthetic Control (Chapter 6.2)
│
├─ Self-selected by units with confounding, but instrument exists? ► IV (Chapter 6.4)
│
├─ Need the week-by-week path of an effect? ───────────────► Pre-trend and gap plots (Chapters 6.1 and 6.2)
│
└─ Need a posterior interval for one of these designs? ────► Bayesian quasi-experiments (Chapter 6.5)
```

### Key Concepts

#### Potential Outcomes Framework

Causal inference relies on the **potential outcomes framework**. For each unit, define two potential outcomes:

- $Y_i(1)$ — outcome if the unit receives treatment
- $Y_i(0)$ — outcome if the unit does not receive treatment

The individual causal effect is $\tau_i = Y_i(1) - Y_i(0)$. The **fundamental problem of causal inference** is that we only ever observe one of these — the factual outcome. The other is the **counterfactual**. Every quasi-experimental method is a strategy for estimating this missing counterfactual.

#### Common Estimands

Different methods target different estimands (the quantity we're trying to estimate):

| Estimand | Definition | Business Question |
|---|---|---|
| **ATE** | $E[Y(1) - Y(0)]$ | Average effect if *everyone* were treated |
| **ATT** | $E[Y(1) - Y(0) \mid D=1]$ | Effect on the units that actually got the intervention (most common in rollout evaluations) |
| **LATE** | Effect on *compliers* | Effect among those whose treatment status was moved by an instrument |
| **CATE** | $E[Y(1) - Y(0) \mid X=x]$ | Effect for a specific segment — fuel for targeting |

Being explicit about the estimand prevents stakeholder confusion: a synthetic control estimates the ATT for one treated market; an IV estimates a LATE that may differ sharply from the ATE.

#### Identifying Assumptions

Every quasi-experimental method buys identification with an assumption. These assumptions are **untestable** — they are claims about unobservable counterfactuals. The craft of quasi-experimentation is:

1. Choosing the assumption that best matches the known facts about *how* the intervention was assigned
2. Attacking that assumption aggressively with falsification tests (placebo tests, pre-trend checks, etc.)
3. Being transparent about limitations with stakeholders

### Strengths and Limitations

**Strengths of Quasi-Experiments:**

- **Applicability**: Can be applied to existing data when RCTs are impossible
- **External validity**: Often use real-world contexts rather than artificial experimental settings
- **Cost-effectiveness**: Leverage data already collected for other purposes
- **Ethical acceptability**: Avoid withholding potentially beneficial treatments

**Limitations:**

- **Assumption dependence**: Results hinge on untestable identifying assumptions
- **Design beats estimation**: The highest-leverage move is often designing better rollouts (e.g., staggering, holding out controls)
- **Local effects**: Many methods identify effects for specific subpopulations (e.g., units near a cutoff for RDD, compliers for IV)
- **Susceptibility to confounding**: Unobserved confounders can still bias estimates if assumptions fail

- You have **structural variation** in treatment assignment (geographic rollouts, thresholds, instruments)
- You **cannot assume unconfoundedness** (treatment assignment depends on unobservables)
- You need to evaluate **historical interventions** that were not designed as experiments

### Practical Guidance

When conducting quasi-experimental analysis in business settings:

1. **Start with the assignment mechanism**: Understand exactly how and why treatment was assigned
2. **Choose the right method**: Match the method to the assignment mechanism, not your preference
3. **Triangulate**: When possible, use multiple methods with different assumptions — agreement strengthens conclusions
4. **Attack assumptions**: Run falsification tests (placebo dates, placebo units, pre-trend checks)
5. **Be transparent**: Report the design, assumption, and estimand — not just the number
6. **Design better next time**: Use insights to improve future rollouts (stagger adoption, hold out controls, randomize encouragement)

PyCausalML fits the estimator. CausalPy is the data source. When CausalPy is installed, `causalpy.load_data` reads the packaged file. Otherwise the notebook reads the same CSV from that GitHub directory.

Later notebooks analyze these CausalPy datasets:

| CausalPy name | File | Design in this series | Study |
|---|---|---|---|
| `banks` | `banks.csv` | Difference-in-differences | Richardson and Troost; Angrist and Pischke, *Mastering Metrics* |
| `california_prop99` | `california_prop99.csv` | Synthetic control | Abadie, Diamond, and Hainmueller (2010), Proposition 99 |
| `drinking` | `drinking.csv` | Regression discontinuity | Minimum legal drinking age |
| `schoolReturns` | `schoolingReturns.csv` | Instrumental variables | Card (1995), college proximity |
| `risk` | `AJR2001.csv` | Instrumental variables | Acemoglu, Johnson, and Robinson (2001) |
| `covid` | `deaths_and_temps_england_wales.csv` | Interrupted time series | Monthly deaths, England and Wales |


````python
CAUSALPY_FILES = {
    "banks": "banks.csv",
    "covid": "deaths_and_temps_england_wales.csv",
    "drinking": "drinking.csv",
    "risk": "AJR2001.csv",
    "schoolReturns": "schoolingReturns.csv",
    "california_prop99": "california_prop99.csv",
}

def load_causalpy(name: str) -> pd.DataFrame:
    """Load a CausalPy example dataset.

    Uses causalpy.load_data when that package is installed. Otherwise reads
    the same CSV from the CausalPy GitHub data directory.
    """
    try:
        import causalpy as cp
        frame = cp.load_data(name)
        print(f"Loaded {name!r} with causalpy.load_data. Shape {frame.shape}.")
        return frame
    except ImportError:
        url = (
            "https://raw.githubusercontent.com/pymc-labs/CausalPy/main/causalpy/data/"
            + CAUSALPY_FILES[name]
        )
        frame = pd.read_csv(url)
        print(f"CausalPy is not installed. Loaded {name!r} from")
        print(url)
        print("Install the package with: pip install 'pycausalml[causalpy]'")
        print(f"Shape {frame.shape}.")
        return frame

catalog = []
for name in ["banks", "california_prop99", "drinking", "schoolReturns", "risk", "covid"]:
    frame = load_causalpy(name)
    catalog.append({"dataset": name, "rows": len(frame), "columns": frame.shape[1]})
print(pd.DataFrame(catalog).to_string(index=False))
````
