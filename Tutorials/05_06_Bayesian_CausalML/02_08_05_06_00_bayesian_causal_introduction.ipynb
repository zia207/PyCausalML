## Overview

### Why Bayesian Causal Inference?

Bayesian methods offer several advantages for causal inference:

1. **Explicit Uncertainty Quantification**: Bayesian methods provide full posterior distributions for causal effects, not just point estimates and confidence intervals. This is crucial for decision-making under uncertainty.

2. **Incorporation of Prior Knowledge**: When data is limited or we have domain expertise, Bayesian methods allow us to encode prior beliefs and update them with data.

3. **Flexibility**: Bayesian models can handle complex hierarchical structures, missing data, and non-standard likelihoods that are difficult for frequentist methods.

4. **Propagating Uncertainty**: In multi-stage causal analyses (e.g., propensity score estimation followed by outcome modeling), Bayesian methods naturally propagate uncertainty through all stages.

5. **Small Sample Performance**: With informative priors, Bayesian methods can produce reliable estimates even with small sample sizes.

### The Bayesian Framework

In Bayesian statistics, we treat unknown parameters as random variables with probability distributions. The key components are:

- **Prior Distribution**: $p(\theta)$ — our beliefs about parameters before seeing data
- **Likelihood**: $p(\text{data} \mid \theta)$ — how likely the data is given parameters
- **Posterior Distribution**: $p(\theta \mid \text{data}) \propto p(\text{data} \mid \theta) \times p(\theta)$ — updated beliefs after seeing data

For causal inference, the parameters of interest are typically treatment effects (ATE, ATT, CATE, etc.). The posterior distribution gives us a full picture of our uncertainty about these effects.

### Bayesian vs. Frequentist Causal Inference

| Aspect | Frequentist | Bayesian |
|---|---|---|
| **Parameters** | Fixed unknown constants | Random variables with distributions |
| **Uncertainty** | Confidence intervals (long-run coverage) | Credible intervals (probability statements) |
| **Prior Knowledge** | Cannot incorporate naturally | Explicitly encoded in priors |
| **Small Samples** | Relies on asymptotics | Can use informative priors |
| **Computation** | Often faster | Requires MCMC/variational inference |
| **Interpretation** | "If we repeated this experiment..." | "Given the data, the probability is..." |

### When to Use Bayesian Causal Methods

Bayesian causal inference is particularly valuable when:

- **Sample sizes are small**: Informative priors can stabilize estimates
- **Prior knowledge exists**: You have domain expertise to encode
- **Uncertainty quantification is critical**: Decision-makers need probability statements
- **Models are complex**: Hierarchical structures, missing data, or non-standard likelihoods
- **Sequential analysis**: You want to update beliefs as new data arrives
- **Propagating uncertainty**: Multi-stage analyses where uncertainty should flow through

### Key Concepts

#### Prior Distributions

Priors encode our beliefs before seeing data. Common choices:

- **Weakly informative priors**: Wide distributions that let data speak (e.g., Normal(0, 10))
- **Informative priors**: Narrow distributions based on domain knowledge
- **Hierarchical priors**: Priors on priors for multi-level models

#### Posterior Distributions

The posterior is our updated belief after seeing data. For causal effects, we get:

- **Point estimates**: Posterior mean or median
- **Uncertainty intervals**: Credible intervals (e.g., 95% highest posterior density)
- **Probability statements**: P(effect > 0) = 0.95

#### Markov Chain Monte Carlo (MCMC)

MCMC algorithms (e.g., NUTS in PyMC) sample from the posterior distribution when analytical solutions are intractable. Key diagnostics:

- **Convergence**: R-hat statistics near 1
- **Effective sample size**: Sufficient independent samples
- **Trace plots**: Visual inspection of chains

### Advantages in Business Contexts

In business settings, Bayesian methods are particularly valuable because:

1. **Natural communication**: "There's a 95% probability the lift is between 3% and 7%" is more intuitive than frequentist confidence intervals

2. **Sequential decision-making**: As new data arrives (e.g., ongoing A/B tests), we can update our posteriors incrementally

3. **Incorporating business knowledge**: Historical results, expert judgment, and business constraints can be encoded as priors

4. **Risk assessment**: Posterior distributions enable Value at Risk (VaR) and other risk metrics

5. **Small experiments**: Pilot programs and early-stage tests often have limited data — Bayesian methods handle this gracefully

## Simple Example: Bayesian Treatment Effect Estimation

Let's start with a simple example: estimating the average treatment effect in a randomized experiment.

````python
# Simulate a simple A/B test
np.random.seed(42)
n_control = 100
n_treatment = 100

# True effect: treatment increases outcome by 5 units
true_effect = 5

# Control group: mean 100, std 15
y_control = np.random.normal(100, 15, n_control)

# Treatment group: mean 105, std 15
y_treatment = np.random.normal(100 + true_effect, 15, n_treatment)

print(f"Control mean: {y_control.mean():.2f}")
print(f"Treatment mean: {y_treatment.mean():.2f}")
print(f"Observed difference: {y_treatment.mean() - y_control.mean():.2f}")
````

Now let's estimate the treatment effect using Bayesian methods:

````python
with pm.Model() as model:
    # Priors for group means (weakly informative)
    mu_control = pm.Normal("mu_control", mu=100, sigma=20)
    mu_treatment = pm.Normal("mu_treatment", mu=100, sigma=20)
    
    # Prior for standard deviation (half-normal)
    sigma = pm.HalfNormal("sigma", sigma=20)
    
    # Treatment effect (difference in means)
    tau = pm.Deterministic("tau", mu_treatment - mu_control)
    
    # Likelihood for control group
    pm.Normal("y_control", mu=mu_control, sigma=sigma, observed=y_control)
    
    # Likelihood for treatment group
    pm.Normal("y_treatment", mu=mu_treatment, sigma=sigma, observed=y_treatment)
    
    # Sample from posterior
    trace = pm.sample(2000, tune=1000, chains=4, target_accept=0.95)
````

Let's examine the posterior distribution of the treatment effect:

````python
# Posterior summary
az.summary(trace, var_names=["tau"], hdi_prob=0.95)
````

````python
# Plot posterior
az.plot_posterior(trace, var_names=["tau"], hdi_prob=0.95)
plt.title("Posterior Distribution of Treatment Effect")
plt.tight_layout()
plt.show()
````

The posterior gives us:

- **Point estimate**: Posterior mean of τ
- **Credible interval**: 95% HDI (highest density interval)
- **Probability statements**: P(τ > 0) from the posterior

This is much richer information than a single number with a confidence interval!

## Model Diagnostics

Before trusting our results, we should check MCMC diagnostics:

````python
# Check convergence (R-hat should be close to 1)
print("R-hat statistics:")
print(az.rhat(trace))

# Trace plots
az.plot_trace(trace, var_names=["tau"])
plt.tight_layout()
plt.show()
````

Good MCMC diagnostics:

- **R-hat < 1.01**: Chains have converged
- **Trace plots**: Look like "fuzzy caterpillars" (good mixing)
- **Effective sample size**: Large enough for stable estimates

## Incorporating Prior Knowledge

One of Bayesian methods' strengths is incorporating prior knowledge. Let's see how informative priors affect the estimate:

````python
with pm.Model() as model_informative:
    # Informative priors based on domain knowledge
    # We believe the effect is around 3 with uncertainty ±2
    mu_control = pm.Normal("mu_control", mu=100, sigma=10)
    mu_treatment = pm.Normal("mu_treatment", mu=103, sigma=10)
    
    sigma = pm.HalfNormal("sigma", sigma=15)
    
    tau = pm.Deterministic("tau", mu_treatment - mu_control)
    
    pm.Normal("y_control", mu=mu_control, sigma=sigma, observed=y_control)
    pm.Normal("y_treatment", mu=mu_treatment, sigma=sigma, observed=y_treatment)
    
    trace_informative = pm.sample(2000, tune=1000, chains=4, target_accept=0.95)

# Compare weakly informative vs informative priors
print("Weakly informative priors:")
print(az.summary(trace, var_names=["tau"], hdi_prob=0.95))

print("\nInformative priors:")
print(az.summary(trace_informative, var_names=["tau"], hdi_prob=0.95))
````

With informative priors, the posterior is "shrunk" toward the prior. This is particularly useful with small samples or when we have strong domain knowledge.

### How uncertainty is represented

Every numeric estimator uses one of two posteriors.

**Conjugate Gaussian posterior.** The conditional mean is linear, and the prior on the slopes is a weak ridge: a small penalty, with the intercept left almost unpenalized. The coefficient posterior is then normal. Its mean is the ridge estimate. Its covariance is the residual variance times the inverse of the penalized Gram matrix. Credible intervals are quantiles of draws from that normal. Difference-in-differences, regression discontinuity, interrupted time series, each equation of a linear-Gaussian network, naive adjustment, and each equation of the time-series model use this posterior.

**Bayesian bootstrap.** A draw reweights the sample with a Dirichlet $(1,\ldots,1)$ vector and refits the estimator. The collection of refits is the posterior. Instrumental variables, synthetic control, and the cross-sectional structural model use it. A two-stage estimator is refit from scratch under each weight vector, so uncertainty from the first stage is inside the interval.

`summary()` returns:

| Key | Meaning |
|---|---|
| `estimate` | Point fit of the causal contrast (ridge estimate or unweighted estimator) |
| `posterior_mean` | Mean of the stored draws |
| `ci_lower`, `ci_upper` | 2.5% and 97.5% quantiles of `draws_` |
| `std` | Standard deviation of the draws |
| `rope_probability` | Included when the constructor received `rope=(low, high)`: the share of draws inside that region of practical equivalence |

The draws stay on `draws_`. They are omitted from `summary()` so a printed summary stays small.

A discrete network is the exception worth remembering. `DiscreteBayesianNetwork.ate` contrasts interventional expectations under the posterior-mean conditional probability tables. It returns that contrast, the treated mean, and the control mean. It does not return a quantile interval. `query` is the method that returns the interventional distribution.

### From a graph to an interval

1. Write the causal contrast. For a binary intervention this is $E[Y \mid do(T=1)] - E[Y \mid do(T=0)]$. In a design it is the parameter that equals that contrast: the DiD interaction, the jump at a cutoff, the IV coefficient.
2. Write the graph, or name the design assumption (parallel trends, continuity at the cutoff, an exclusion restriction).
3. Read off an adjustment set, a frontdoor set, or an instrument. If the graph has none, the effect is not identified from the observed variables.
4. Fit the structural equations or the design regression.
5. Report the posterior of the contrast.

Observing the treatment and intervening on it are different queries. $P(Y \mid T=1)$ still travels along any open backdoor. $P(Y \mid do(T=1))$ deletes the arrows into $T$ and then propagates. The discrete-network example prints both numbers from one fitted model.

### 1. Causal graphs and identification

`CausalGraph` stores a directed acyclic graph. The edge `(parent, child)` means the parent is a direct cause of the child. Every common cause you are willing to use has to appear as a node: once the graph is drawn, identification treats it as causally sufficient. A cycle raises `ValueError`.

**d-separation** is the test the rest of the rules are built from.

| Graph | Edges | What the criterion says |
|---|---|---|
| Chain | A → B → C | A and C are d-separated by B |
| Fork | A ← B → C | A and C are d-separated by B |
| Collider | A → B ← C | A and C are d-separated when nothing is conditioned on. Conditioning on B opens the path |

**Backdoor criterion.** A set $Z$ identifies the effect of $T$ on $Y$ when all three of the following hold.

- $Z$ contains neither $T$ nor $Y$.
- No member of $Z$ is a descendant of $T$. A descendant can sit on a causal path, or it can be a collider that opens a non-causal path.
- In the graph with the arrows out of $T$ removed, $Z$ d-separates $T$ and $Y$. Removing those arrows leaves the non-causal paths, which is what an adjustment set has to block.

`minimal_adjustment_set(treatment, outcome)` returns one smallest valid set, as a sorted list of names. `adjustment_sets` returns every inclusion-minimal set among the empty set, the parents of treatment, and subsets of the non-descendants up to `max_size`. If no observed set works, `minimal_adjustment_set` raises `ValueError`.

**Frontdoor criterion.** A mediator set $M$ identifies the effect when four things hold.

- $M$ is non-empty and contains neither treatment nor outcome.
- Every directed path from treatment to outcome passes through $M$.
- Treatment and $M$ have no open backdoor. They are d-separated once the arrows out of treatment are removed.
- Every backdoor from $M$ to the outcome is blocked by treatment.

`frontdoor_sets(treatment, outcome)` returns those mediator sets. This is the identification result used when the confounder is latent and a mediator on the causal path is observed. The graph names the mediator. A discrete network can then evaluate the interventional distribution with `query`.

**Instruments.** `instruments(treatment, outcome)` lists observed nodes that have a directed path into the treatment and are d-separated from the outcome after the treatment node is removed. A node with its own arrow into the outcome fails that test and is left out. A latent confounder of treatment and outcome does not, by itself, remove an instrument: if the instrument has no arrow into the outcome and no connection to the confounder except through treatment, exclusion still holds.

`adjacency()` returns a DataFrame with `A.loc[child, parent] = 1`. `parents`, `children`, `ancestors`, and `descendants` expose the same graph for inspection.


````python
chain = CausalGraph([("A", "B"), ("B", "C")])
fork = CausalGraph([("B", "A"), ("B", "C")])
collider = CausalGraph([("A", "B"), ("C", "B")])

print("Chain: A ⊥ C | B ", chain.d_separated("A", "C", "B"))
print("Chain: A ⊥ C     ", chain.d_separated("A", "C"))
print("Fork:  A ⊥ C | B ", fork.d_separated("A", "C", "B"))
print("Collider: A ⊥ C  ", collider.d_separated("A", "C"))
print("Collider: A ⊥ C | B", collider.d_separated("A", "C", "B"))

# Engagement C confounds a loyalty offer T and spend Y.
confounded = CausalGraph([("C", "T"), ("C", "Y"), ("T", "Y")])
print("Backdoor set:", confounded.minimal_adjustment_set("T", "Y"))
print("Empty set is valid:", confounded.satisfies_backdoor("T", "Y", []))
print("Adjusting for C is valid:", confounded.satisfies_backdoor("T", "Y", ["C"]))

# A mediator M of the offer is a descendant. It is not a backdoor set.
mediated = CausalGraph([("C", "T"), ("C", "Y"), ("T", "M"), ("M", "Y")])
print("Adjustment set with a mediator in the graph:", mediated.minimal_adjustment_set("T", "Y"))
print("Adjusting for the mediator is valid:", mediated.satisfies_backdoor("T", "Y", ["M"]))

# Latent U confounds T and Y. M sits on the only causal path.
frontdoor = CausalGraph([("U", "T"), ("U", "Y"), ("T", "M"), ("M", "Y")])
print("Frontdoor sets:", frontdoor.frontdoor_sets("T", "Y"))

instrument = CausalGraph([("Z", "T"), ("T", "Y")])
leaked = CausalGraph([("Z", "T"), ("T", "Y"), ("Z", "Y")])
print("Instrument when Z is excluded from Y:", instrument.instruments("T", "Y"))
print("Instrument when Z also points at Y:", leaked.instruments("T", "Y"))

confounded.adjacency()
````

The prints are the identification checklist in miniature.

- In the chain and the fork, the middle node blocks the path. In the collider, the path is blocked until you condition on the collision node. That is why a descendant of treatment is a dangerous control.
- With $C \rightarrow T$, $C \rightarrow Y$, and $T \rightarrow Y$, the smallest backdoor set is $\{C\}$. The empty set leaves the fork open.
- Adding the mediator $T \rightarrow M \rightarrow Y$ does not put $M$ into the adjustment set. `satisfies_backdoor` rejects $\{M\}$ because $M$ is a descendant of treatment and lies on the causal path.
- In the frontdoor graph the confounder $U$ is latent, so it cannot be an adjustment covariate. The mediator $M$ still satisfies the frontdoor criterion, and the effect remains identified.
- $Z$ is an instrument when the only way it reaches $Y$ is through $T$. An extra arrow $Z \rightarrow Y$ removes it from `instruments`.

The adjacency matrix uses the package convention `A.loc[child, parent] = 1`: the row is the effect, the column is the cause.


### 2. Linear-Gaussian Bayesian network

`GaussianBayesianNetwork` turns the graph into a structural equation for every node:

$$
X_j = a_j + \sum_{k \in \mathrm{pa}(j)} \beta_{jk} X_k + \varepsilon_j,
\qquad \varepsilon_j \sim \mathcal{N}(0, \sigma_j^2).
$$

`fit(data)` expects a data frame with one column per node and estimates a separate Gaussian posterior for each equation. Parents are taken from the graph, so a column that the graph does not name as a parent stays out of that equation.

`ate(treatment, outcome, treated=1, control=0)` is the posterior of

$$
E[Y \mid do(T = \texttt{treated})] - E[Y \mid do(T = \texttt{control})].
$$

In a linear system that contrast equals the total effect times `treated - control`. The total effect sums the products of structural coefficients along every directed path from treatment to outcome. Each posterior draw perturbs every equation's coefficients, then recomputes the path sum, so the interval includes uncertainty in the confounder equations as well as in the outcome equation.

`sample(n, do={"T": 1.0})` draws from the posterior-mean equations in topological order. A node listed in `do` is clamped to that value and its parents are ignored for that node. The difference between the outcome mean under `do(T=1)` and under `do(T=0)` is a simulation estimate of the same total effect.

Call `summary()` after `ate`. The draws of the total effect are on `draws_`.

The simulation below is a loyalty offer. Prior engagement $C$ pushes customers into the offer and also shifts spend. The structural effect of the offer on spend is $+2$. A regression of spend on the offer alone leaves the engagement fork open, so its slope mixes the offer with engagement. The network includes both arrows out of $C$ and recovers the structural slope.


````python
rng = np.random.default_rng(0)
n = 4000
engagement = rng.normal(size=n)
offer = 0.8 * engagement + rng.normal(scale=0.5, size=n)
spend = 2.0 * offer - 1.1 * engagement + rng.normal(scale=0.6, size=n)
frame = pd.DataFrame({"C": engagement, "T": offer, "Y": spend})

naive = np.linalg.lstsq(
    np.column_stack([np.ones(n), offer]), spend, rcond=None
)[0][1]

network = GaussianBayesianNetwork(
    [("C", "T"), ("C", "Y"), ("T", "Y")],
    n_draws=500,
    random_state=0,
)
network.fit(frame)
ate = network.ate("T", "Y", treated=1.0, control=0.0)

print(f"Naive slope of spend on offer: {naive:.3f}")
print("Network posterior of the total effect:")
display(pd.Series(ate))

do_offer = network.sample(5000, do={"T": 1.0}, random_state=1)
do_control = network.sample(5000, do={"T": 0.0}, random_state=2)
print(
    "Simulated E[Y | do(T=1)] - E[Y | do(T=0)]:",
    f"{do_offer['Y'].mean() - do_control['Y'].mean():.3f}",
)

fig, ax = plt.subplots(figsize=(7.2, 4.0))
ax.hist(network.draws_, bins=40, density=True, color="C0", alpha=0.85)
low, high = np.quantile(network.draws_, [0.025, 0.975])
ax.axvspan(low, high, color="C0", alpha=0.18, label="95% credible interval")
ax.axvline(2.0, color="black", linestyle="--", linewidth=1.2, label="structural effect = 2")
ax.set_xlabel("total effect of the offer on spend")
ax.set_ylabel("posterior density")
ax.set_title(r"Gaussian Bayesian network: $E[Y \mid do(T=1)] - E[Y \mid do(T=0)]$")
ax.legend(frameon=False)
fig.tight_layout()
plt.show()
````

The naive slope sits away from 2 because engagement is inside both the offer and spend. The network posterior concentrates on the structural value 2: the backdoor through engagement is in the graph, so it is removed from the offer equation and the spend equation before the path product is formed. The `do` samples land on the same contrast. Clamping `T` cuts the arrow from engagement into the offer, which is the intervention the total effect refers to.


### 3. Discrete Bayesian network

`DiscreteBayesianNetwork` is the same graph with categorical conditionals. For a node with states $\mathcal{X}$ and Dirichlet concentration `alpha` (the default is 1, one pseudo-count per state), the posterior mean of a conditional probability is

$$
\hat P(x \mid \mathrm{pa})
  = \frac{n(x, \mathrm{pa}) + \alpha}{n(\mathrm{pa}) + \alpha \, |\mathcal{X}|}.
$$

`fit` builds one such table for every parent configuration that appears in the data. `cpd(node)` returns that table as a data frame.

`query(variables, evidence=None, do=None)` returns marginal distributions.

- `do={node: value}` replaces that node's conditional with a point mass at `value`. Parents of the node no longer affect it. This is an intervention.
- `evidence={node: value}` conditions on the observed value and still uses the node's conditional. This is association.
- A node cannot appear in both maps.

The joint is enumerated exactly when its size is at most `exact_limit` cells (default 8192). The result then has `"method": "exact"`. Larger joints use likelihood weighting and `"method": "likelihood_weighting"`. Probabilities are keyed by the original state labels.

`ate(treatment, outcome, treated=1, control=0)` is defined when the outcome labels are numeric. It subtracts $E[Y \mid do(T=\mathrm{control})]$ from $E[Y \mid do(T=\mathrm{treated})]$. String labels need `query` rather than `ate`.

Two processes follow. Both use binary nodes, so the exact enumerator applies.

**A chain.** $T \rightarrow M \rightarrow Y$, with $P(M=1 \mid T) = 0.2 + 0.6\,T$ and $P(Y=1 \mid M) = 0.1 + 0.8\,M$. There is no backdoor. The interventional contrast is

$$
\begin{aligned}
E[Y \mid do(T=1)] &= 0.8 \times 0.9 + 0.2 \times 0.1 = 0.74, \\
E[Y \mid do(T=0)] &= 0.2 \times 0.9 + 0.8 \times 0.1 = 0.26,
\end{aligned}
$$

so the average treatment effect is $0.48$. Conditioning on $T$ agrees with intervening on $T$, because nothing confounds the chain.

**A confounder.** $C$ causes both $T$ and $Y$, and $T$ also causes $Y$, with

$$
P(Y=1 \mid T, C) = 0.1 + 0.2\,T + 0.6\,C
$$

and $P(C=1) = 1/2$. Under an intervention the confounder stays at its marginal, so

$$
E[Y \mid do(T=1)] = 0.1 + 0.2 + 0.6 \times 0.5 = 0.60.
$$

Conditioning on $T=1$ over-samples $C=1$, because the confounder is what produced treatment. $P(Y=1 \mid T=1)$ is therefore larger than the interventional probability. That gap is the confounding the graph removes.


````python
rng = np.random.default_rng(0)
n = 8000

# Chain: the offer changes redemption, and redemption changes purchase.
treatment = rng.binomial(1, 0.5, size=n)
mediator = rng.binomial(1, np.where(treatment == 1, 0.8, 0.2))
outcome = rng.binomial(1, np.where(mediator == 1, 0.9, 0.1))
chain_frame = pd.DataFrame({"T": treatment, "M": mediator, "Y": outcome})

chain_net = DiscreteBayesianNetwork(
    [("T", "M"), ("M", "Y")], alpha=0.5, random_state=0
)
chain_net.fit(chain_frame)
print("Chain ATE (structural value 0.48):")
display(pd.Series(chain_net.ate("T", "Y", treated=1, control=0)))
print("CPD of the mediator:")
display(chain_net.cpd("M"))

# Confounded offer: engagement drives both take-up and purchase.
confounder = rng.binomial(1, 0.5, size=n)
treatment = rng.binomial(1, np.where(confounder == 1, 0.9, 0.1))
probability = 0.1 + 0.2 * treatment + 0.6 * confounder
outcome = rng.binomial(1, probability)
confounded_frame = pd.DataFrame({"C": confounder, "T": treatment, "Y": outcome})

confounded_net = DiscreteBayesianNetwork(
    [("C", "T"), ("C", "Y"), ("T", "Y")], alpha=0.5, random_state=0
)
confounded_net.fit(confounded_frame)
interventional = confounded_net.query("Y", do={"T": 1})
conditional = confounded_net.query("Y", evidence={"T": 1})
print("Query method:", interventional["method"])
print("P(Y=1 | do(T=1)):", f"{interventional['distributions']['Y'][1]:.3f}")
print("P(Y=1 | T=1):    ", f"{conditional['distributions']['Y'][1]:.3f}")
print("Confounded-graph ATE:")
display(pd.Series(confounded_net.ate("T", "Y", treated=1, control=0)))
````

On the chain, the fitted average treatment effect sits on 0.48, and the mediator's conditional probability table reproduces the two assignment probabilities, about 0.2 and 0.8. On the confounded graph, $P(Y=1 \mid do(T=1))$ sits on 0.60, while $P(Y=1 \mid T=1)$ is pulled upward by the customers whose engagement both triggered the offer and raised purchase. Reporting the conditional probability as the effect of the offer would credit the offer for engagement. The `do` query leaves engagement at its population mix.

The printed ATE is about 0.20, which is the coefficient on $T$ in the probability formula. The control interventional mean is $E[Y \mid do(T=0)] = 0.1 + 0.6 \times 0.5 = 0.40$, and $0.60 - 0.40 = 0.20$.


### 4. Adjustment and instrumental variables from the graph

The network classes estimate every structural equation. Two thinner estimators skip the equations you do not need and read the identifying set from a `CausalGraph`.

`NaiveAdjustmentRegressor(graph, treatment, outcome)` calls `minimal_adjustment_set` and fits the Bayesian outcome regression

$$
Y = a + \tau \, T + Z^{\top}\gamma + \varepsilon.
$$

Under the graph and a linear outcome model, $\tau$ is the average treatment effect. `summary()` includes `adjustment_set`. `predict(data)` evaluates the fitted mean. The posterior on $\tau$ is the conjugate Gaussian posterior from the ridge regression.

`NaiveIVRegressor(graph, treatment, outcome)` calls `instruments` and runs Bayesian two-stage least squares. `summary()` includes `instruments`. If `instruments` returns an empty list, `fit` raises `ValueError`.

The graph you hand the estimator is the identifying assumption.

- For adjustment, draw every confounder you can condition on. The class uses the smallest backdoor set and ignores mediators that fail the backdoor criterion.
- For an instrument, draw $Z \rightarrow T \rightarrow Y$. A latent confounder of $T$ and $Y$ can stay off the drawn graph: the assumption encoded by that picture is that $Z$ reaches $Y$ only through $T$, and that $Z$ is independent of the latent confounder. Adding an arrow $Z \rightarrow Y$ is an exclusion failure, `instruments` returns `[]`, and the regressor refuses to fit.

The adjustment example reuses the loyalty-offer frame from the Gaussian network, where the structural effect is 2 and the adjustment set is engagement. The instrumental-variable example is a separate draw: an encouragement $Z$ shifts take-up, a latent shock confounds take-up and the outcome, and the structural effect is 1.5. Ordinary least squares of the outcome on take-up is biased. Two-stage least squares, with $Z$ read off the graph, targets 1.5.


````python
adjusted = NaiveAdjustmentRegressor(confounded, "T", "Y", n_draws=800, random_state=0)
adjusted.fit(frame)
print("Backdoor regression on the loyalty-offer frame:")
display(pd.Series(adjusted.summary()))

rng = np.random.default_rng(2)
n_iv = 3000
encouragement = rng.normal(size=n_iv)
latent = rng.normal(size=n_iv)
takeup = 0.9 * encouragement + 0.7 * latent + rng.normal(scale=0.2, size=n_iv)
response = 1.5 * takeup + 0.8 * latent + rng.normal(scale=0.3, size=n_iv)
iv_frame = pd.DataFrame({"Z": encouragement, "T": takeup, "Y": response})

ols_iv = np.linalg.lstsq(
    np.column_stack([np.ones(n_iv), takeup]), response, rcond=None
)[0][1]
print(f"OLS slope of response on take-up: {ols_iv:.3f}")

iv_graph = CausalGraph([("Z", "T"), ("T", "Y")])
print("Instruments on the exclusion graph:", iv_graph.instruments("T", "Y"))
print(
    "Instruments after adding Z → Y:",
    CausalGraph([("Z", "T"), ("T", "Y"), ("Z", "Y")]).instruments("T", "Y"),
)

iv_model = NaiveIVRegressor(iv_graph, "T", "Y", n_draws=200, random_state=0)
iv_model.fit(iv_frame)
print("Bayesian IV posterior (structural effect 1.5):")
display(pd.Series(iv_model.summary()))
````

### 5. Structural causal model for an endogenous treatment

`BayesianStructuralSCM` writes the treatment and the outcome as a pair of structural assignments:

$$
\begin{aligned}
T &= a_T + X_T^{\top}\beta_T + V, \\
Y &= \alpha \, T + a_Y + X_Y^{\top}\beta_Y + U,
\qquad \mathrm{Corr}(U, V) = \rho.
\end{aligned}
$$

When $\rho \neq 0$, the treatment residual $V$ is inside $T$. Ordinary least squares of $Y$ on $T$ and the covariates is then biased for $\alpha$, because the regression can partial out $X$ and still leave $V$ in the treatment. The residual correlation is unobserved confounding.

Exclusion restrictions split the columns of $X$:

| Role | Argument | Where it enters |
|---|---|---|
| Instrument | `instruments` | Treatment equation only |
| Outcome covariate | `outcome_only` | Outcome equation only |
| Confounder | every remaining column | Both equations |

Column indices or column names both work. `fit(X, treatment, y, instruments=..., outcome_only=...)` runs two-stage least squares and stores a Bayesian bootstrap of both stages. `identified_by_` is `"instrumental_variable"` when instruments are passed. Omit `instruments` only when treatment is conditionally ignorable given $X$: the fit is then a Bayesian outcome regression and `identified_by_` is `"conditional_ignorability"`.

Under $do(T = t)$ the outcome equation is evaluated at that value of $t$. The treatment residual is an estimation device. It is not part of the intervened assignment. Three prediction methods use that intervened equation:

- `predict(X, treatment)` returns the mean outcome at the supplied treatment values.
- `predict_interval(X, treatment, level=0.95)` adds a pointwise credible band from the coefficient draws.
- `potential_outcomes(X)` returns $Y(0)$ and $Y(1)$. With a homogeneous $\alpha$, the row-wise contrast equals the structural effect.

`simulate_structural_scm(n, alpha, rho)` builds a known version of this design. Five standardized covariates are returned in `X`: columns 0 and 1 are confounders, columns 2 and 3 are instruments, and column 4 predicts the outcome only. Treatment and outcome errors are bivariate normal with correlation `rho`. The structural effect is `alpha`. The dictionary also returns `instruments`, `outcome_only`, and `feature_names`, which can be passed straight into `fit`.


````python
data = simulate_structural_scm(n=2500, alpha=3.0, rho=0.6, random_state=0)
ols_design = np.column_stack(
    [np.ones(len(data["y"])), data["treatment"], data["X"]]
)
ols_alpha = np.linalg.lstsq(ols_design, data["y"], rcond=None)[0][1]

scm = BayesianStructuralSCM(n_draws=200, random_state=0)
scm.fit(
    data["X"],
    data["treatment"],
    data["y"],
    instruments=data["instruments"],
    outcome_only=data["outcome_only"],
    feature_names=data["feature_names"],
)

print(f"OLS coefficient on treatment: {ols_alpha:.3f}")
print(f"Identification: {scm.identified_by_}")
print("Structural posterior (true alpha = 3):")
display(pd.Series(scm.summary()))

y0, y1 = scm.potential_outcomes(data["X"][:5])
print("Y(1) - Y(0) on five rows (homogeneous effect):")
print(np.round(y1 - y0, 3))
````

Ordinary least squares is pulled off 3 by the correlated residuals. The structural fit, which keeps the instruments out of the outcome equation and the outcome-only covariate out of the treatment equation, puts the posterior on 3. Because $\alpha$ is homogeneous, $Y(1) - Y(0)$ is the same number on every row, and that number is the posterior point fit of $\alpha$.


### 6. Structural time series

`BayesianTimeSeriesSCM` is the structural model on a multivariate series. Contemporaneous parents form a DAG, passed as `edges`. Each equation also contains its own lags and the lags of those parents, up to `lag` (the default is 1). An empty edge list is a system of separate autoregressions. Every equation has its own Gaussian coefficient posterior.

Two interventions answer two different questions.

**A hard intervention** replaces an equation for the whole horizon. The clamped series does not consult its parents or its lags. `counterfactual(horizon, interventions, baseline=None, history=None)` subtracts a baseline rollout from the intervened rollout.

- `interventions` maps a variable to a scalar or to a horizon-length sequence.
- `baseline=None` is the observational forecast from `forecast`.
- `baseline={"X": 0.0}` together with `interventions={"X": 1.0}` is the sustained contrast of $do(X=1)$ against $do(X=0)\`.

The returned dictionary holds `names`, `counterfactual`, `baseline`, `effect`, `effect_lower`, and `effect_upper`. The last two are pointwise 2.5% and 97.5% quantiles across coefficient draws.

**An impulse** adds a shock to the structural residual for one period, then lets every equation run as fitted. `impulse_response(variable, shock=1, horizon=10)` returns `response`, `baseline`, and `effect`. A clamp answers "what if this series is held at a value." An impulse answers "what if this series is hit once and then reacts."

`summary(node)` returns that equation's `feature_names`, posterior-mean `coef`, and residual variance `sigma2`. Names look like `X[t]`, `Y[t-1]`, and `X[t-1]`: the `[t]` term is the contemporaneous parent, and the lag terms are the node's own past and its parents' past.

`history` has one column per series, in the order of `names_` (topological order), and at least `lag` rows. Leave it out to start from the end of the fitted sample. The example below starts from a zero history so the contrast is the effect of the intervention rather than a continuation of a particular sample path.

The data-generating process is a weekly demand series. A driver $X$ is autoregressive with coefficient 0.4. Demand $Y$ loads on the current driver with coefficient 0.8 and on its own lag with coefficient 0.2. There is no direct lag from $X$ into $Y$ beyond what the contemporaneous arrow and $Y$'s own lag induce.


````python
rng = np.random.default_rng(0)
n_weeks = 1200
driver = np.zeros(n_weeks)
demand = np.zeros(n_weeks)
for t in range(1, n_weeks):
    driver[t] = 0.4 * driver[t - 1] + rng.normal(0.0, 0.5)
    demand[t] = 0.8 * driver[t] + 0.2 * demand[t - 1] + rng.normal(0.0, 0.3)

ts = BayesianTimeSeriesSCM(edges=[("X", "Y")], lag=1, n_draws=150, random_state=0)
ts.fit(pd.DataFrame({"X": driver, "Y": demand}))
equation = ts.summary("Y")
print("Posterior mean of the demand equation (structural loading on X[t] is 0.8):")
display(pd.Series(equation["coef"], index=equation["feature_names"], name="coef"))

history = np.zeros((ts.lag, len(ts.names_)))
impulse = ts.impulse_response("X", shock=1.0, horizon=8, history=history)
y_index = impulse["names"].index("Y")
print(f"One-period response of Y: {impulse['effect'][0, y_index]:.3f}")

contrast = ts.counterfactual(
    24, {"X": 1.0}, baseline={"X": 0.0}, history=history
)
y_index = contrast["names"].index("Y")
horizon = np.arange(contrast["effect"].shape[0])

fig, axes = plt.subplots(1, 2, figsize=(10.5, 4.0))
axes[0].plot(impulse["effect"][:, y_index], marker="o", color="C0")
axes[0].axhline(0.0, color="0.5", linewidth=0.8)
axes[0].set_title("Impulse: unit shock to X, then the system reacts")
axes[0].set_xlabel("weeks after the shock")
axes[0].set_ylabel("effect on Y")

axes[1].plot(horizon, contrast["effect"][:, y_index], color="C0", label="posterior mean")
axes[1].fill_between(
    horizon,
    contrast["effect_lower"][:, y_index],
    contrast["effect_upper"][:, y_index],
    color="C0",
    alpha=0.2,
    label="95% credible band",
)
axes[1].set_title(r"Hard intervention: $do(X=1)$ against $do(X=0)$")
axes[1].set_xlabel("weeks from a zero history")
axes[1].set_ylabel("effect on Y")
axes[1].legend(frameon=False)
fig.tight_layout()
plt.show()
````

The demand equation's coefficient on `X[t]` recovers the contemporaneous loading 0.8, and the coefficient on `Y[t-1]` recovers the own lag 0.2. The coefficient on `X[t-1]` stays near 0, which is the lag the simulation did not include.

The impulse plot jumps by that contemporaneous loading in week 0 and then decays: after the shock, $X$ and $Y$ follow their fitted dynamics. The hard-intervention plot rises toward the steady gap between holding the driver at 1 and holding it at 0. With loading 0.8, own lag 0.2, and no parent lag, that gap is $0.8 / (1 - 0.2) = 1$. The credible band is the posterior uncertainty in the coefficients, rolled forward under both interventions.


### 7. Quasi-experimental posteriors

Five classes encode a study design directly, for the case where the design itself is the identification argument and a unit-level graph is unnecessary. Notebooks 6.2–6.4 rebuild difference-in-differences, interrupted time series, and instrumental variables in PyMC, which is the tool to open when you want a hierarchical likelihood, a custom prior, or a posterior predictive check. The classes here are the ones to call from `pycausalml` when the design in the table is the design you have.

| Class | Call | Causal parameter |
|---|---|---|
| `BayesianDifferenceInDifferences` | `fit(y, treated, post)` | $\tau$ in $Y = a + b\,\mathrm{treated} + c\,\mathrm{post} + \tau\,\mathrm{treated}\times\mathrm{post}$. This is the average treatment effect on the treated under parallel trends. `treated` and `post` are 0/1 indicators. |
| `BayesianRegressionDiscontinuity` | `fit(running, y, treatment=None)` | Jump at `cutoff` in the local linear design $1,\ (R-c),\ D,\ (R-c)D$. Rows with $\lvert R-c\rvert$ above `bandwidth` are dropped when a bandwidth is set. If `treatment` is omitted, $D = 1\{R \ge c\}$. The number of rows kept is `n_window_`. |
| `BayesianInterruptedTimeSeries` | `fit(time, y, intervention_time)` | Immediate level shift in $Y = a + b\,\mathrm{time} + \tau\,\mathrm{post} + d\,\mathrm{time\ since}$. The slope change $d$ is stored on `slope_change_`. |
| `BayesianInstrumentalVariable` | `fit(y, treatment, instrument, X=None)` | Second-stage coefficient on fitted treatment. `instrument` may have several columns. `X` holds covariates that enter both stages. Draws are a Bayesian bootstrap of both stages. |
| `BayesianSyntheticControl` | `fit(control_pre, treated_pre, control_post, treated_post)` | Mean post-period gap for one treated unit. Donor weights are non-negative and sum to one. `control_pre` has shape `(n_pre, n_donors)`. The post-period gap path is `effect_path_`; `effect_` is its mean. |

The linear designs and the instrumental-variable class accept `rope=(low, high)`. The summary then includes `rope_probability`, the posterior probability that the effect sits inside an interval you would treat as negligible for the decision.

The fit below is a two-group, two-period panel with a true interaction of 1.5. The region of practical equivalence is $(-0.25, 0.25)$: effects inside that band would not change the decision.


````python
rng = np.random.default_rng(0)
n_panel = 400
treated = np.repeat([0.0, 1.0], n_panel // 2)
post = np.tile([0.0, 1.0], n_panel // 2)
outcome = (
    1.0
    + 0.2 * treated
    + 0.3 * post
    + 1.5 * treated * post
    + rng.normal(scale=0.25, size=n_panel)
)

did = BayesianDifferenceInDifferences(
    n_draws=1000, rope=(-0.25, 0.25), random_state=0
)
did.fit(outcome, treated, post)
print("Coefficient names:", did.feature_names_)
print("Posterior of the treated × post interaction (true value 1.5):")
display(pd.Series(did.summary()))
````

`treated_x_post` is the causal parameter. Its posterior sits on 1.5, and `rope_probability` is the share of draws inside $\pm 0.25$. A near-zero rope probability is the Bayesian statement that the ATT is large enough to act on, under the parallel-trends assumption built into the interaction.

The other four classes return the same summary keys from the calls in the table. Pass `random_state` so the draws repeat, and keep `n_draws` at a few hundred: these intervals are quantiles of a normal or of a bootstrap, and they settle quickly.


### Choosing an estimator

| You have | Call |
|---|---|
| An observed confounder and a linear outcome | `GaussianBayesianNetwork` or `NaiveAdjustmentRegressor` |
| Categorical variables and a query under `do` | `DiscreteBayesianNetwork` |
| A latent confounder and an observed mediator that meets the frontdoor criterion | `CausalGraph.frontdoor_sets`, then `DiscreteBayesianNetwork.query` |
| An instrument named by the graph | `NaiveIVRegressor` |
| An instrument and covariates, without building a graph | `BayesianInstrumentalVariable` |
| Exclusion restrictions on a covariate matrix | `BayesianStructuralSCM` |
| Two groups and a known adoption date, with parallel trends | `BayesianDifferenceInDifferences` |
| A cutoff on a running variable | `BayesianRegressionDiscontinuity` |
| One series and an intervention date | `BayesianInterruptedTimeSeries` |
| One treated unit and a donor pool | `BayesianSyntheticControl` |
| A multivariate series, and either a hard intervention or a one-period shock | `BayesianTimeSeriesSCM` |

Use the PyMC notebooks when the likelihood is hierarchical, the prior is something other than a weak ridge, or you want a posterior predictive check of the outcome distribution. Use this module when the graph or the design above is the model and the deliverable is a posterior interval for the identified contrast.
